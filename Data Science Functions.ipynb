# Bonus latihan (2 Parameter)

## Latihan 1


```python
def perkenalan(nama,umur):
    print("Nama:",nama)
    print("Umur:",umur)
perkenalan("Budi",23)
```

## Latihan 2


```python
def data_produk(nama,harga):
    print("Produk:",nama)
    print("Harga:",harga)
data_produk("Mouse",150000)
```

## Latihan 3

```python
def data_buku(judul,penulis):
    print("Judul:",judul)
    print("Penulis:",penulis)
data_buku("Belajar Python", "Andi")
```

```python
def data_mahasiswa(nama,jurusan):
    print("Nama:",nama)
    print("jurusan:",jurusan)
data_mahasiswa("Rina","Teknik Informatika")
```
